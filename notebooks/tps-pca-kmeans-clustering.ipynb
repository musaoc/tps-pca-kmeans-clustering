# 4- PCA- Dimension Reductionality

```python
#lets scale the data for further use.
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
```

```python
pca = PCA(0.95) 
```

```python
df_pca = pca.fit_transform(df_scaled)
```

```python
df_pca.shape
```

```python
df_pca = pd.DataFrame(df_pca)
df_pca.head()
```

```python
sns.heatmap(df_pca.corr());
```

# 5- Modeling 🤖

## 5-A- Kmeans

**Using elbow method to define the number of clusters 💪**

```python
from yellowbrick.cluster import KElbowVisualizer

model = KMeans()
visualizer = KElbowVisualizer(model, k=(4,12))

visualizer.fit(df_pca)     
visualizer.show()        
```

```python
km = KMeans(n_clusters=7)
y = km.fit_predict(df_pca)
```

```python
y
```

```python
df_og['Predicted']=y
```

```python
submission=df_og.loc[:,['id','Predicted']]
submission.shape
```

```python
submission.head()
```

```python
submission.to_csv('submission.csv',index=False)
```

```python
### Using Gausian mixture with same number of clusters
from sklearn.mixture import GaussianMixture
gmm = GaussianMixture(n_components = 7)
 
# Fit the GMM model for the dataset
gmm.fit(df_pca)
 
# Assign a label to each sample
labels = gmm.predict(df_pca)
df['Clusters']= labels
```

```python
df.head()
```

# 7- Submission ✨

```python
submission['id']=df.index
submission['Predicted']=df.loc[:,['Clusters']]
submission.shape
```
